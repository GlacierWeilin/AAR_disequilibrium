```python
import numpy as np
import pandas as pd
```

```python
file_path = '/scratch/k10/wy2165/Loibl/'
AAR = []
for i in range(1,7):
    filename = file_path + 'AAR_1985_2021_Loibl_' + str(i) + '.csv';
    df = pd.read_csv(filename).iloc[:, 1:]
    AAR.append(df)
AAR = pd.concat(AAR, axis=1)
AAR = AAR.T
AAR.index.name = 'RGIId'
AAR.columns = np.arange(1985, 2021)

AAR.to_csv('/g/data/rd53/wy2165/disequilibrium/data/AAR_1985_2021_Loibl.csv')
```
