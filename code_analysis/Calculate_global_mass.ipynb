```python
import numpy as np
import pandas as pd
```

```python
regions = np.arange(1,20,1)
outpath = '/g/data/rd53/wy2165/disequilibrium/pygem_oggm/'
```

### Median a

```python
all_dfs = []
for region in regions:

    if region in {1, 3, 4, 5, 7, 9, 17, 19}:
        fn = outpath + f'PyGEM_regional_mass_{region:02d}_median_a.csv'
    else:
        fn = outpath + f'PyGEM_regional_mass_{region:02d}_median_k.csv'

    df_reg = pd.read_csv(fn, index_col=0)

    all_dfs.append(df_reg)

df_all = pd.concat(all_dfs, ignore_index=True)

keep_cols = ['gcm', 'period_scenario', 'temp_ch_ipcc']

sum_cols = [
    'area_rgi',
    'mass_2020',
    'area_steady',
    'volume_steady',
    'mass_steady',
]

df_sum = (
    df_all
    .groupby(keep_cols, as_index=False)[sum_cols]
    .sum()
)

df_sum['mass_remaining'] = df_sum['mass_steady'] / df_sum['mass_2020'] * 100
```

```python
df_sum.to_csv(outpath + 'PyGEM_global_mass_median_a.csv', index=False)
```

### Median k

```python
df_sum.to_csv(outpath + 'PyGEM_global_mass_median_k.csv', index=False)
```
