```python
import numpy as np
import pandas as pd
```

```python
glaciermip3_path = '/g/data/rd53/wy2165/disequilibrium/GlacierMIP3/'
pygem_path = '/g/data/rd53/wy2165/disequilibrium/pygem_oggm/'

glaciermip3 = pd.read_csv(
    glaciermip3_path + 'fig1b_scatter_by_temp_model.csv'
)

this_study = pd.read_csv(
    pygem_path + 'PyGEM_global_mass_median_a.csv'
)
```

```python
glaciermip3_mass = glaciermip3[
    ['gcm', 'period_scenario', 'temp_ch_ipcc', 'mass_percent_rel_2020']
].copy()

glaciermip3_mass = glaciermip3_mass.rename(
    columns={'mass_percent_rel_2020': 'mass_remaining'}
)

this_study_mass = this_study[
    ['gcm', 'period_scenario', 'temp_ch_ipcc', 'mass_remaining']
].copy()

this_study_mass_match = this_study_mass.merge(
    glaciermip3_mass[
        ['gcm', 'period_scenario', 'temp_ch_ipcc']
    ].drop_duplicates(),
    on=['gcm', 'period_scenario'],
    how='inner',
    suffixes=('_this_study', '')
)

this_study_mass_match = this_study_mass_match[
    ['gcm', 'period_scenario', 'temp_ch_ipcc', 'mass_remaining']
].copy()

this_study_mass_era5 = this_study_mass[
    np.isclose(this_study_mass['temp_ch_ipcc'], 1.2)
].copy()

this_study_mass_era5['gcm'] = 'era5'
this_study_mass_era5['period_scenario'] = '2014-2023'

this_study_mass_era5 = this_study_mass_era5[
    ['gcm', 'period_scenario', 'temp_ch_ipcc', 'mass_remaining']
].copy()

this_study_mass = pd.concat(
    [this_study_mass_match, this_study_mass_era5],
    ignore_index=True
)

all_mass = pd.concat(
    [glaciermip3_mass, this_study_mass],
    ignore_index=True
)
```

```python
all_mass.to_csv(
    glaciermip3_path + 'GlacierMIP3_add_this_study.csv',
    index=False
)
```

```python
mass_median = (
    all_mass
    .groupby('temp_ch_ipcc', as_index=False)['mass_remaining']
    .median()
)
```

```python
mass_median.to_csv(
    glaciermip3_path + 'GlacierMIP3_add_this_study_median.csv',
    index=False
)
```
