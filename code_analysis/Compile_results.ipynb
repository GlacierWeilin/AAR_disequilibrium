```python
import os
import glob

import pandas as pd
```

```python
region = 5
file_path = f'/scratch/k10/wy2165/PyGEM/oggm_gdirs/None/{region:02d}/'
```

```python
csv_files = sorted(glob.glob(os.path.join(file_path, f'glacier_statistics_PyGEM_{region:02d}_batch_*.csv')))
df_list = [pd.read_csv(f) for f in csv_files]
merged_df = pd.concat(df_list, ignore_index=True)
merged_df.to_csv(os.path.join(file_path, f'glacier_statistics_PyGEM_{region:02d}.csv'), index=False)
```

```python
gcm = 'ERA5'
csv_files = sorted(glob.glob(os.path.join(file_path, f'glacier_disequilibrium_PyGEM_{region:02d}_{gcm}_batch_*.csv')))
df_list = [pd.read_csv(f) for f in csv_files]
merged_df = pd.concat(df_list, ignore_index=True)
merged_df.to_csv(os.path.join(file_path, f'glacier_disequilibrium_PyGEM_{region:02d}_{gcm}.csv'), index=False)
```

### Compile disequilibrium simulations

```python
gcms = ['GFDL-ESM4', 'IPSL-CM6A-LR', 'MPI-ESM1-2-HR', 'MRI-ESM2-0', 'UKESM1-0-LL']
ssps = ['ssp126', 'ssp370', 'ssp585']
```

```python
for gcm in gcms:
    for ssp in ssps:
        csv_files = sorted(glob.glob(os.path.join(file_path, f'glacier_disequilibrium_PyGEM_{region:02d}_{gcm}_{ssp}_batch_*.csv')))
        df_list = [pd.read_csv(f) for f in csv_files]
        merged_df = pd.concat(df_list, ignore_index=True)
        merged_df.to_csv(os.path.join(file_path, f'glacier_disequilibrium_PyGEM_{region:02d}_{gcm}_{ssp}.csv'), index=False)
```
